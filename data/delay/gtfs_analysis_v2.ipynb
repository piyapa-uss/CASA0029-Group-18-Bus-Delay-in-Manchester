# GTFS — Study area filter & OSM network prep (v2)

Two-phase setup notebook for the downstream GTFS routing pipeline,
optimised for batch processing across multiple dates.

**v2 changes (vs. v1):**
- §5.9 is now a *unified-static* driver: every date's GTFS is streamed into
  a single union before any expensive work runs. Node-snapping
  (`ox.nearest_nodes`) and A* route geometry construction happen **once**
  across the union, not once per date. For ≥3 dates this typically halves
  total runtime.
- A static-GTFS hash is reported per date so you can see at a glance whether
  the feeds are identical or genuinely changing.
- RAM-conscious: per-date GTFS dicts are dropped immediately after being
  folded into the union, and per-date `vehicles` / `snapped` frames are
  freed as soon as they're consumed.

**Phase A — Study-area filter.** Given a polygon, find every GTFS route that
has at least one stop inside it on any of the selected dates. Output is a
list of `route_id`s. Full routes are kept (not just the in-area portion) so
downstream geometry / arrival reconstruction sees complete context.

**Phase B — OSM network prep.** Download the OSM `drive`, `tram`, and `walk`
graphs once (resumable, tiled), then crop on demand to the bbox you need.
Avoids re-downloading per-region or per-day.

**Phase C — GTFS routing & arrival reconstruction.** Build per-direction
route geometries, snap RT vehicle pings, interpolate / extrapolate stop
arrival times. New in v2: shared static state across all dates.

## Inputs
- `STUDY_AREA_DIR` — polygon files; all unioned into one geometry.
- `DATA_DIR` — root of `YYYYMMDD/itm_<region>/` GTFS folders.
- `DATES` — one or more `YYYYMMDD` strings; route_ids are unioned.

## Outputs
- `data_checkpoint/` — Phase A persistence (route_ids, polygon, bbox).
- `output/routes_<date>.gpkg` + `.arrivals.csv` — Phase C results.
- `output/cache/route_geoms_v3_dir_unified.pickle` — shared route geometries.

# Phase A — Study-area filter

## 2. Load and dissolve the study-area polygon

```python
import geopandas as gpd
import pandas as pd
from shapely.ops import unary_union

POLY_EXTS = {".gpkg", ".shp", ".geojson", ".json"}

poly_files = [p for p in STUDY_AREA_DIR.iterdir()
              if p.is_file() and p.suffix.lower() in POLY_EXTS]
if not poly_files:
    raise FileNotFoundError(f"No polygon files found in {STUDY_AREA_DIR}")

print(f"Found {len(poly_files)} polygon file(s):")
parts = []
for f in poly_files:
    g = gpd.read_file(f)
    print(f"  {f.name}: {len(g)} features, CRS={g.crs}")
    # Reproject to WGS84 to match GTFS stops
    if g.crs is None:
        raise ValueError(f"{f} has no CRS set — cannot reproject safely.")
    g = g.to_crs("EPSG:4326")
    parts.append(g)

all_polys = gpd.GeoDataFrame(pd.concat(parts, ignore_index=True),
                             crs="EPSG:4326")
study_area_geom = unary_union(all_polys.geometry.values)  # dissolve
gdf_study_area  = gpd.GeoDataFrame(geometry=[study_area_geom], crs="EPSG:4326")

print(f"\nDissolved into {len(gdf_study_area)} geometry "
      f"({study_area_geom.geom_type}); area ≈ "
      f"{gdf_study_area.to_crs('EPSG:27700').area.iloc[0] / 1e6:.1f} km²")
gdf_study_area.explore()
```

## 3. Identify routes that pass the study area

### 3.1 For each date: stops inside the polygon → routes that serve them

```python
def load_gtfs_stops(gtfs_dir: Path) -> gpd.GeoDataFrame:
    """Load stops.txt as a GeoDataFrame in WGS84."""
    stops = pd.read_csv(gtfs_dir / "stops.txt", dtype=str, low_memory=False)
    stops["stop_lat"] = pd.to_numeric(stops["stop_lat"], errors="coerce")
    stops["stop_lon"] = pd.to_numeric(stops["stop_lon"], errors="coerce")
    stops = stops.dropna(subset=["stop_lat", "stop_lon"])
    return gpd.GeoDataFrame(
        stops,
        geometry=gpd.points_from_xy(stops["stop_lon"], stops["stop_lat"]),
        crs="EPSG:4326",
    )


def routes_passing_study_area(gtfs_dir: Path, study_geom):
    """Return (route_ids set, gdf_stops_in_area, gdf_all_stops_on_selected_routes)."""
    gdf_stops = load_gtfs_stops(gtfs_dir)

    # Spatial filter: stops inside the study area
    in_area_mask = gdf_stops.within(study_geom)
    gdf_in = gdf_stops[in_area_mask].copy()
    if gdf_in.empty:
        return set(), gdf_in, gdf_stops.iloc[0:0].copy()

    # trip_id → route_id
    trips = pd.read_csv(gtfs_dir / "trips.txt", dtype=str, low_memory=False)
    trip_to_route = dict(zip(trips["trip_id"], trips["route_id"]))

    # stop_times → trip_ids that touch any in-area stop
    in_area_stops = set(gdf_in["stop_id"])
    stop_times = pd.read_csv(gtfs_dir / "stop_times.txt", dtype=str,
                             low_memory=False, usecols=["trip_id", "stop_id"])
    in_area_trips = set(stop_times.loc[
        stop_times["stop_id"].isin(in_area_stops), "trip_id"])

    selected_routes = {trip_to_route[t] for t in in_area_trips
                       if t in trip_to_route}

    # All stops on those routes (full coverage, not just in-area ones)
    selected_trips = {t for t, r in trip_to_route.items() if r in selected_routes}
    selected_stop_ids = set(stop_times.loc[
        stop_times["trip_id"].isin(selected_trips), "stop_id"])
    gdf_route_stops = gdf_stops[gdf_stops["stop_id"].isin(selected_stop_ids)].copy()
    gdf_route_stops["in_study_area"] = gdf_route_stops["stop_id"].isin(in_area_stops)

    return selected_routes, gdf_in, gdf_route_stops


per_date = {}
for date in DATES:
    gtfs_dir = DATA_DIR / date / REGION
    if not gtfs_dir.is_dir():
        print(f"  [WARN] Missing GTFS folder: {gtfs_dir} — skipping")
        continue
    routes, gdf_in, gdf_rs = routes_passing_study_area(gtfs_dir, study_area_geom)
    per_date[date] = {"routes": routes,
                      "stops_in_area": gdf_in,
                      "route_stops": gdf_rs}
    print(f"  {date}: {len(gdf_in):,} stops in area, {len(routes):,} routes selected")
```

### 3.2 Union route_ids across dates

```python
selected_route_ids = set().union(*(d["routes"] for d in per_date.values()))
print(f"Union over {len(per_date)} date(s): {len(selected_route_ids):,} unique routes")

# Per-date presence matrix (helpful to spot routes that disappear on some days)
presence = pd.DataFrame({
    date: [rid in d["routes"] for rid in sorted(selected_route_ids)]
    for date, d in per_date.items()
}, index=sorted(selected_route_ids))
presence.index.name = "route_id"
presence["n_dates_present"] = presence.sum(axis=1)
presence.head(20)
```

### 3.3 Inspect routes & stops on a chosen date

```python
# Pick a representative date for visual inspection
INSPECT_DATE = DATES[0]
d = per_date[INSPECT_DATE]

# All stops on selected routes, flagged whether they fall inside the polygon
gdf_selected_routes_stops = d["route_stops"]
gdf_study_stops           = d["stops_in_area"]

print(f"On {INSPECT_DATE}:")
print(f"  stops inside study area : {len(gdf_study_stops):,}")
print(f"  total stops on selected routes (full extent): "
      f"{len(gdf_selected_routes_stops):,}")
print(f"  of those, inside study area: "
      f"{int(gdf_selected_routes_stops['in_study_area'].sum()):,}")
print(f"  outside study area (kept for context): "
      f"{int((~gdf_selected_routes_stops['in_study_area']).sum()):,}")

# Quick visual: study-area polygon + in-area stops + outside-area context stops
m = gdf_study_area.explore(color="#ffcc00", style_kwds={"fillOpacity": 0.2},
                            name="Study area")
m = gdf_selected_routes_stops[~gdf_selected_routes_stops["in_study_area"]].explore(
    m=m, color="#888888", marker_kwds={"radius": 2}, name="Outside (context)")
m = gdf_selected_routes_stops[gdf_selected_routes_stops["in_study_area"]].explore(
    m=m, color="#cc0033", marker_kwds={"radius": 3}, name="Inside study area")
m
```

# Phase B — OSM network prep

## 4. Whole-UK OSM network (one-time download, then crop per run)

Downloading the OSM graph for every per-day bbox is wasteful when the bboxes overlap. Instead we download the whole UK once, cache it tile-by-tile, and crop to the bbox we need at run time.

**Cost:** the UK drive graph is large (rough order: ~1.5 GB pickled, several GB of RAM while loading). Walk is larger still. Tram is small. Run §4.1 once per machine, then §4.3 each time you need a region-specific subgraph.

## 4. OSM network prep

Loading these from disk means **you can restart the kernel and run Phase B/C
without re-doing the per-date GTFS scan**. If a checkpoint is missing, run

### 4.1 Build cropped drive / tram / walk graphs

Three source options — set `OSM_SOURCE` in the config cell below.

| `OSM_SOURCE`    | Mechanism | RAM use | When to use |
|-----------------|-----------|---------|-------------|
| `"osmnx_bbox"`  | Overpass API — fetches only the bbox region | **Low** (~100–500 MB) | **Default — simplest, most RAM-friendly** |
| `"pbf"`         | pyrosm reads local `.pbf`, filters by bbox  | **High** — reads full file into RAM before filtering | Only if Overpass is unavailable / slow |
| `"osmnx_tiled"` | Tiled UK-wide Overpass download, resumable  | High once, low per crop | Only if you need the whole UK graph |

All sources produce the same output: three `MultiDiGraph` objects
(`drive`, `tram`, `walk`) cropped to the requested bbox, cached as pickles
so repeat runs for the same area are instant.


```python
import time
import pickle
import hashlib
import math
from pathlib import Path
import networkx as nx
import osmnx as ox

# ── Config ───────────────────────────────────────────────────────────────────
OSM_SOURCE      = "osmnx_bbox"   # "osmnx_bbox" | "pbf" | "osmnx_tiled"
OSM_NETWORK_DIR = Path(r"D:\2026_03-Bus_Project\data_input\osm_network")
PBF_FILE        = None           # None = autodetect first *.pbf in OSM_NETWORK_DIR
GRAPH_CACHE_DIR = Path(r"D:\2026_03-Bus_Project\output\osm_graphs_uk")
GRAPH_CACHE_DIR.mkdir(parents=True, exist_ok=True)

# osmnx tiled settings (only when OSM_SOURCE="osmnx_tiled")
UK_BBOX      = (-8.65, 49.85, 1.80, 60.90)
TILE_DEG     = 1.0
TILE_OVERLAP = 0.02

TRAM_FILTER = '["railway"~"tram|light_rail"]'

ox.settings.cache_folder = str(GRAPH_CACHE_DIR / "osmnx_cache")
ox.settings.use_cache    = True


# ─────────────────────────────────────────────────────────────────────────────
# Shared helpers
# ─────────────────────────────────────────────────────────────────────────────

def _save_pickle(obj, path: Path):
    """Atomic write — never leaves a half-written file on Ctrl-C."""
    tmp = path.with_suffix(path.suffix + ".tmp")
    with open(tmp, "wb") as fh:
        pickle.dump(obj, fh, protocol=pickle.HIGHEST_PROTOCOL)
    tmp.replace(path)


def _empty_graph():
    G = nx.MultiDiGraph()
    G.graph["crs"] = "epsg:4326"
    return G


def _bbox_cache_key(name: str, bbox) -> str:
    raw = f"{name}_{bbox[0]:.4f}_{bbox[1]:.4f}_{bbox[2]:.4f}_{bbox[3]:.4f}"
    return hashlib.md5(raw.encode()).hexdigest()[:10]


def _load_cache(name: str, bbox):
    """Return cached graph if it exists, else None."""
    pkl = GRAPH_CACHE_DIR / f"{name}_{_bbox_cache_key(name, bbox)}.pickle"
    if not pkl.exists():
        return None, pkl
    size_mb = pkl.stat().st_size / 1_048_576
    print(f"  {name}: loading cache ({size_mb:.0f} MB) …", end=" ", flush=True)
    t0 = time.monotonic()
    with open(pkl, "rb") as fh:
        G = pickle.load(fh)
    print(f"{len(G.nodes):,} nodes, {len(G.edges):,} edges "
          f"({time.monotonic()-t0:.1f}s)")
    return G, pkl


# ─────────────────────────────────────────────────────────────────────────────
# Source A — osmnx_bbox  (default, low RAM)
# Queries Overpass for exactly the bbox — no whole-UK graph, minimal RAM.
# ─────────────────────────────────────────────────────────────────────────────

def _load_osmnx_bbox(name: str, bbox, **graph_from_bbox_kwargs) -> nx.MultiDiGraph:
    G, pkl = _load_cache(name, bbox)
    if G is not None:
        return G

    minlon, minlat, maxlon, maxlat = bbox
    print(f"  {name}: querying Overpass for bbox "
          f"lon {minlon:.3f}–{maxlon:.3f}, lat {minlat:.3f}–{maxlat:.3f} …",
          flush=True)
    t0 = time.monotonic()
    try:
        G = ox.graph_from_bbox(bbox=bbox, simplify=True,
                               **graph_from_bbox_kwargs)
    except Exception as exc:
        print(f"    [WARN] {exc} — returning empty graph")
        G = _empty_graph()
    print(f"    {len(G.nodes):,} nodes, {len(G.edges):,} edges "
          f"({time.monotonic()-t0:.1f}s)")
    _save_pickle(G, pkl)
    return G


# ─────────────────────────────────────────────────────────────────────────────
# Source B — pbf via pyrosm  (local file, but still high RAM on large pbf)
# ─────────────────────────────────────────────────────────────────────────────

def _resolve_pbf() -> Path:
    global PBF_FILE
    if PBF_FILE is not None:
        return Path(PBF_FILE)
    candidates = sorted(OSM_NETWORK_DIR.glob("*.pbf"))
    if not candidates:
        raise FileNotFoundError(
            f"No *.pbf found in {OSM_NETWORK_DIR}")
    return candidates[0]


def _pyrosm_to_nx(nodes_gdf, edges_gdf) -> nx.MultiDiGraph:
    import numpy as np
    G = nx.MultiDiGraph()
    G.graph["crs"] = "epsg:4326"
    if nodes_gdf is not None and len(nodes_gdf):
        for nid, x, y in zip(nodes_gdf["id"].to_numpy(),
                              nodes_gdf.geometry.x.to_numpy(),
                              nodes_gdf.geometry.y.to_numpy()):
            G.add_node(int(nid), x=float(x), y=float(y))
    if edges_gdf is None or len(edges_gdf) == 0:
        return G
    us     = edges_gdf["u"].to_numpy()
    vs     = edges_gdf["v"].to_numpy()
    lens   = (edges_gdf["length"].to_numpy()
              if "length" in edges_gdf.columns else np.full(len(edges_gdf), np.nan))
    geoms  = edges_gdf.geometry.to_numpy()
    oneways = (edges_gdf["oneway"].fillna("no").astype(str).str.lower().to_numpy()
               if "oneway" in edges_gdf.columns else ["no"] * len(edges_gdf))
    for u, v, L, geom, ow in zip(us, vs, lens, geoms, oneways):
        u, v  = int(u), int(v)
        attrs = {"length": float(L) if L == L else None, "geometry": geom}
        G.add_edge(u, v, **attrs)
        if ow not in {"yes", "true", "1", "-1"}:
            G.add_edge(v, u, **attrs)
    return G


def _load_pbf(name: str, bbox,
              network_type: str = None,
              custom_filter: dict = None) -> nx.MultiDiGraph:
    G, pkl = _load_cache(name, bbox)
    if G is not None:
        return G

    from pyrosm import OSM
    pbf = _resolve_pbf()
    minlon, minlat, maxlon, maxlat = bbox
    print(f"  {name}: parsing {pbf.name} (bbox-filtered) …", flush=True)
    t0  = time.monotonic()
    osm = OSM(str(pbf), bounding_box=[minlon, minlat, maxlon, maxlat])
    if custom_filter is not None:
        nodes, edges = osm.get_network(nodes=True, network_type="all",
                                       extra_attributes=["oneway"])
        mask = None
        for k, vs in custom_filter.items():
            m = edges[k].isin(vs) if k in edges.columns else None
            mask = m if mask is None else (mask | m)
        edges = edges[mask].copy() if mask is not None else edges.iloc[0:0]
        used  = set(edges["u"]).union(edges["v"])
        nodes = nodes[nodes["id"].isin(used)].copy()
    else:
        nodes, edges = osm.get_network(nodes=True, network_type=network_type,
                                       extra_attributes=["oneway"])
    print(f"    parsed {len(nodes):,} nodes, {len(edges):,} edges "
          f"({time.monotonic()-t0:.1f}s) — building graph …")
    t0 = time.monotonic()
    G  = _pyrosm_to_nx(nodes, edges)
    print(f"    graph built ({time.monotonic()-t0:.1f}s)")
    _save_pickle(G, pkl)
    return G


# ─────────────────────────────────────────────────────────────────────────────
# Source C — tiled osmnx (whole-UK, resumable)
# ─────────────────────────────────────────────────────────────────────────────

def _iter_tiles(bbox, tile_deg, overlap):
    minlon, minlat, maxlon, maxlat = bbox
    i = 0; lon = minlon
    while lon < maxlon:
        j = 0; lat = minlat
        while lat < maxlat:
            tb = (max(lon - overlap, minlon), max(lat - overlap, minlat),
                  min(lon + tile_deg + overlap, maxlon),
                  min(lat + tile_deg + overlap, maxlat))
            yield (i, j), tb
            lat += tile_deg; j += 1
        lon += tile_deg; i += 1


def _merge_tiles(name, tile_dir, final_pkl, verbose=True):
    Gs = []
    for p in sorted(tile_dir.glob("tile_*.pickle")):
        try:
            with open(p, "rb") as fh:
                Gs.append(pickle.load(fh))
        except Exception as exc:
            print(f"    [WARN] {p.name}: {exc}")
    G = nx.compose_all(Gs) if Gs else _empty_graph()
    G.graph["crs"] = "epsg:4326"
    _save_pickle(G, final_pkl)
    if verbose:
        print(f"    merged: {len(G.nodes):,} nodes, {len(G.edges):,} edges "
              f"({final_pkl.stat().st_size/1_048_576:.0f} MB)")
    return G


def _load_osmnx_tiled(name: str, **kw) -> nx.MultiDiGraph:
    final_pkl  = GRAPH_CACHE_DIR / f"{name}_uk.pickle"
    tile_dir   = GRAPH_CACHE_DIR / f"{name}_tiles"
    tile_dir.mkdir(parents=True, exist_ok=True)
    tiles      = list(_iter_tiles(UK_BBOX, TILE_DEG, TILE_OVERLAP))
    n_expected = len(tiles)
    n_have     = len(list(tile_dir.glob("tile_*.pickle")))
    if final_pkl.exists() and n_have == n_expected:
        print(f"  {name}: cache complete — loading …", end=" ", flush=True)
        t0 = time.monotonic()
        with open(final_pkl, "rb") as fh:
            G = pickle.load(fh)
        print(f"{len(G.nodes):,} nodes ({time.monotonic()-t0:.1f}s)")
        return G
    print(f"  {name}: {n_expected} tiles, {n_have} already on disk")
    new_this_run = 0
    for k, ((i, j), tb) in enumerate(tiles, start=1):
        pkl = tile_dir / f"tile_{i:03d}_{j:03d}.pickle"
        if pkl.exists():
            continue
        t0 = time.monotonic()
        try:
            G_tile = ox.graph_from_bbox(bbox=tb, simplify=True, **kw)
        except Exception as exc:
            print(f"    [{k}/{n_expected}] tile {i:03d}_{j:03d} EMPTY ({exc})")
            G_tile = _empty_graph()
        _save_pickle(G_tile, pkl)
        new_this_run += 1
        print(f"    [{k}/{n_expected}] tile {i:03d}_{j:03d} "
              f"{len(G_tile.nodes):>7,}n  {time.monotonic()-t0:5.1f}s")
        if new_this_run % 10 == 0:
            _merge_tiles(name, tile_dir, final_pkl, verbose=False)
    return _merge_tiles(name, tile_dir, final_pkl)


# ─────────────────────────────────────────────────────────────────────────────
# Public entry point
# ─────────────────────────────────────────────────────────────────────────────

def get_networks(bbox: tuple) -> dict:
    """
    Return {"drive": G, "tram": G, "walk": G, "none": None} for `bbox`.

    Source is controlled by OSM_SOURCE:
      "osmnx_bbox"  — query Overpass for this bbox only (default, low RAM)
      "pbf"         — parse local .pbf filtered to bbox (high RAM warning)
      "osmnx_tiled" — whole-UK tiled download then in-memory crop
    """
    if OSM_SOURCE == "osmnx_bbox":
        drive = _load_osmnx_bbox("drive", bbox, network_type="drive")
        tram  = _load_osmnx_bbox("tram",  bbox, custom_filter=TRAM_FILTER,
                                 retain_all=True)
        walk  = _load_osmnx_bbox("walk",  bbox, network_type="walk")

    elif OSM_SOURCE == "pbf":
        drive = _load_pbf("drive", bbox, network_type="driving")
        tram  = _load_pbf("tram",  bbox,
                          custom_filter={"railway": ["tram", "light_rail"]})
        walk  = _load_pbf("walk",  bbox, network_type="walking")

    elif OSM_SOURCE == "osmnx_tiled":
        G_drive = _load_osmnx_tiled("drive", network_type="drive")
        G_tram  = _load_osmnx_tiled("tram",  custom_filter=TRAM_FILTER,
                                    retain_all=True)
        G_walk  = _load_osmnx_tiled("walk",  network_type="walk")
        drive   = crop_graph_to_bbox(G_drive, bbox)
        tram    = crop_graph_to_bbox(G_tram,  bbox)
        walk    = crop_graph_to_bbox(G_walk,  bbox)

    else:
        raise ValueError(f"Unknown OSM_SOURCE={OSM_SOURCE!r}. "
                         f"Use 'osmnx_bbox', 'pbf', or 'osmnx_tiled'.")

    return {"drive": drive, "tram": tram, "walk": walk, "none": None}
```

### 4.2 Crop a cached graph to a bounding box

Pure in-memory filter on node `(lon, lat)` — fast (seconds), allocation-light,
original graph untouched.

The bbox can be padded one of two ways:

| Argument      | Meaning                                          | Default |
|---------------|--------------------------------------------------|---------|
| `buffer_km`   | Uniform distance buffer in **kilometres**        | `5.0`   |
| `buffer_pct`  | Fraction of the bbox span (e.g. `0.1` = 10 %)    | off     |

Pass at most one. With both `None` the bbox is taken as-is. The km buffer is
converted to degrees using the bbox-centre latitude (`1° lat ≈ 111 km`,
`1° lon ≈ 111·cos(lat)`), which is accurate enough for buffers up to a few
hundred km in the UK.

```python
import math


def buffer_bbox(bbox, buffer_km: float = 5.0, buffer_pct: float = None):
    """
    Pad a (min_lon, min_lat, max_lon, max_lat) bbox.

    Exactly one of `buffer_km` (default 5 km) and `buffer_pct` should be set;
    pass the other as None to disable. With both None the bbox is unchanged.
    """
    minlon, minlat, maxlon, maxlat = bbox

    if buffer_km is not None and buffer_pct is not None:
        raise ValueError("Pass either buffer_km or buffer_pct, not both")

    if buffer_km is not None:
        lat_mid = 0.5 * (minlat + maxlat)
        d_lat = buffer_km / 111.0
        d_lon = buffer_km / (111.0 * max(math.cos(math.radians(lat_mid)), 1e-6))
    elif buffer_pct is not None:
        d_lat = (maxlat - minlat) * buffer_pct
        d_lon = (maxlon - minlon) * buffer_pct
    else:
        d_lat = d_lon = 0.0

    return (minlon - d_lon, minlat - d_lat,
            maxlon + d_lon, maxlat + d_lat)


def crop_graph_to_bbox(G, bbox, buffer_km: float = 5.0,
                       buffer_pct: float = None):
    """
    Return the induced subgraph of G on nodes whose (x, y) lie within
    `bbox`, optionally padded by `buffer_km` (default 5 km) or `buffer_pct`.

    Original `G` is not modified. All edge attributes survive.
    """
    minlon, minlat, maxlon, maxlat = buffer_bbox(bbox, buffer_km, buffer_pct)
    keep = [n for n, d in G.nodes(data=True)
            if d.get("x") is not None and d.get("y") is not None
            and minlon <= d["x"] <= maxlon
            and minlat <= d["y"] <= maxlat]
    H = G.subgraph(keep).copy()
    H.graph["crs"] = G.graph.get("crs", "epsg:4326")
    return H


def stops_bbox(gdf_stops):
    """(min_lon, min_lat, max_lon, max_lat) from a stops GeoDataFrame."""
    minx, miny, maxx, maxy = gdf_stops.total_bounds
    return (float(minx), float(miny), float(maxx), float(maxy))
```

# Phase C — GTFS routing & arrival reconstruction

Inlined, direction-aware version of the v3 pipeline (`gtfs_stop_router_v3.py`).

Two fixes vs. v3:

1. **Direction split** — routes are keyed by `(route_id, direction_id)` so
   inbound and outbound trips no longer collapse onto a single LineString.
   v3 picked the longest trip across the whole route, which mixed both
   directions if their stop counts were similar.

2. **Pre-service / post-service extrapolation** — when a stop has only
   pings *after* it (or only *before* it) but at least two of them, we
   linearly extrapolate the arrival time using the velocity of the two
   closest pings. v3 returned NaN here, which dropped the first / last few
   stops on most trips. New `arrival_source` tags: `extrapolated_pre`,
   `extrapolated_post`.

Restricts to the `selected_route_ids` from Phase A, so we only process
routes that pass the study area.


## 5.1 Constants & helpers

```python
import datetime as _dt
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import geopandas as gpd
import networkx as nx
import osmnx as ox
from pyproj import Transformer
from shapely.geometry import LineString, Point
from shapely.ops import transform as shp_transform
import shapely

ROUTE_TYPE_NAMES = {0: "Tram", 3: "Bus", 4: "Ferry", 200: "Coach"}
ROUTE_TYPE_NETWORK = {0: "tram", 3: "drive", 4: "none", 200: "drive"}
WGS84      = "EPSG:4326"
METRIC_CRS = "EPSG:27700"


def make_projectors():
    to_metric = Transformer.from_crs(WGS84, METRIC_CRS, always_xy=True).transform
    to_wgs    = Transformer.from_crs(METRIC_CRS, WGS84, always_xy=True).transform
    return to_metric, to_wgs


def parse_gtfs_time(s):
    """'HH:MM:SS' (incl. >24:00:00) → seconds-since-midnight."""
    try:
        h, m, sc = s.strip().split(":")
        return int(h) * 3600 + int(m) * 60 + int(sc)
    except Exception:
        return np.nan


def auto_utc_offset(feed_ts):
    month = _dt.datetime.utcfromtimestamp(feed_ts).month
    return 3600 if 3 <= month <= 10 else 0


def _norm_dir(d):
    """Coerce direction_id to '0'/'1'; missing → '0'."""
    if d is None or (isinstance(d, float) and np.isnan(d)):
        return "0"
    s = str(d).strip()
    return s if s in {"0", "1"} else "0"
```

## 5.2 Static GTFS loader (direction-aware)

Filters to `selected_route_ids`, then for every `(route_id, direction_id)`
key builds:

* `route_dir_trips_map` — list of trip_ids in this direction
* `route_dir_stops`     — ordered union of stops across those trips
* `route_dir_canonical` — the longest-stop-list trip in this direction (used
  as the spine when stitching the route geometry)


```python
def load_static_gtfs(gtfs_dir: Path, route_id_filter: set = None):
    gtfs_dir = Path(gtfs_dir)
    print(f"  Loading static GTFS from {gtfs_dir} …")

    def read(fname):
        return pd.read_csv(gtfs_dir / fname, dtype=str, low_memory=False)

    trips_df   = read("trips.txt")
    routes_df  = read("routes.txt")
    stops_df   = read("stops.txt")
    stop_times = read("stop_times.txt")

    # Optional restriction to a route_id allowlist (Phase A output)
    if route_id_filter is not None:
        rf = set(map(str, route_id_filter))
        trips_df  = trips_df[trips_df["route_id"].isin(rf)].copy()
        routes_df = routes_df[routes_df["route_id"].isin(rf)].copy()
        kept_trip_ids = set(trips_df["trip_id"])
        stop_times    = stop_times[stop_times["trip_id"].isin(kept_trip_ids)].copy()
        print(f"    after route filter: {len(routes_df):,} routes, "
              f"{len(trips_df):,} trips, {len(stop_times):,} stop_times")

    stops_df["stop_lat"] = pd.to_numeric(stops_df["stop_lat"], errors="coerce")
    stops_df["stop_lon"] = pd.to_numeric(stops_df["stop_lon"], errors="coerce")
    stops_df = stops_df.dropna(subset=["stop_lat", "stop_lon"])
    stop_coords   = (stops_df.set_index("stop_id")[["stop_lon", "stop_lat"]]
                              .apply(tuple, axis=1).to_dict())
    stop_name_map = stops_df.set_index("stop_id")["stop_name"].to_dict()

    stop_times["stop_sequence"] = pd.to_numeric(stop_times["stop_sequence"],
                                                errors="coerce")
    stop_times["arr_sec"]       = stop_times["arrival_time"].apply(parse_gtfs_time)

    trip_stops = (
        stop_times[["trip_id", "stop_id", "stop_sequence"]]
        .sort_values(["trip_id", "stop_sequence"])
        .groupby("trip_id")["stop_id"].apply(list).to_dict()
    )

    if "route_type" in routes_df.columns:
        routes_df["route_type"] = pd.to_numeric(routes_df["route_type"],
                                                errors="coerce").astype("Int64")
    else:
        routes_df["route_type"] = pd.Series([3] * len(routes_df), dtype="Int64")
    route_type_map  = routes_df.set_index("route_id")["route_type"].to_dict()
    route_short_map = routes_df.set_index("route_id")["route_short_name"].to_dict()

    # ── Direction-aware trip → route grouping ───────────────────────────────
    if "direction_id" not in trips_df.columns:
        trips_df["direction_id"] = "0"
    trips_df["direction_id"] = trips_df["direction_id"].apply(_norm_dir)

    trip_dir = dict(zip(trips_df["trip_id"], trips_df["direction_id"]))
    trip_route = trips_df[["trip_id", "route_id", "direction_id"]].drop_duplicates("trip_id")

    route_dir_trips_map = {}
    for _, r in trip_route.iterrows():
        key = (r["route_id"], r["direction_id"])
        route_dir_trips_map.setdefault(key, []).append(r["trip_id"])

    # Ordered union of stops across all trips for each (route, dir)
    route_dir_stops = {}
    route_dir_canonical = {}
    for key, tids in route_dir_trips_map.items():
        seen, seen_set = [], set()
        canon = []
        for tid in tids:
            ts = trip_stops.get(tid, [])
            if len(ts) > len(canon):
                canon = ts
            for sid in ts:
                if sid not in seen_set:
                    seen.append(sid); seen_set.add(sid)
        route_dir_stops[key]     = seen
        route_dir_canonical[key] = canon

    gtfs_trip_ids = set(trips_df["trip_id"])

    print(f"    Trips: {len(trips_df):,}  |  Routes: {len(routes_df):,}  "
          f"|  (route, dir) keys: {len(route_dir_trips_map):,}  "
          f"|  Stops: {len(stops_df):,}")

    return {
        "trips_df":              trips_df,
        "routes_df":             routes_df,
        "stops_df":              stops_df,
        "stop_times":            stop_times,
        "stop_coords":           stop_coords,
        "stop_name_map":         stop_name_map,
        "trip_stops":            trip_stops,
        "trip_dir":              trip_dir,
        "route_dir_trips_map":   route_dir_trips_map,
        "route_dir_stops":       route_dir_stops,
        "route_dir_canonical":   route_dir_canonical,
        "route_type_map":        route_type_map,
        "route_short_map":       route_short_map,
        "gtfs_trip_ids":         gtfs_trip_ids,
    }
```

## 5.3 GTFS-RT snapshot loader

```python
def load_gtfsrt(bin_path, gtfs_trip_ids, routes_df, trips_df):
    from google.transit import gtfs_realtime_pb2

    with open(bin_path, "rb") as f:
        first = f.read(4)
    if not first or first[0] != 0x0A:
        return None, None

    with open(bin_path, "rb") as f:
        feed = gtfs_realtime_pb2.FeedMessage()
        try:
            feed.ParseFromString(f.read())
        except Exception:
            return None, None

    feed_ts = feed.header.timestamp
    rows = []
    for e in feed.entity:
        v = e.vehicle
        if not v.HasField("position") or v.trip.trip_id not in gtfs_trip_ids:
            continue
        rows.append({
            "entity_id":  e.id,
            "trip_id":    v.trip.trip_id,
            "route_id":   v.trip.route_id,
            "vehicle_id": v.vehicle.id,
            "lat":        v.position.latitude,
            "lon":        v.position.longitude,
            "bearing":    v.position.bearing,
            "timestamp":  v.timestamp,
        })
    if not rows:
        return None, feed_ts

    vehicles = pd.DataFrame(rows)
    vehicles = vehicles[(vehicles.lat > 49) & (vehicles.lat < 61) &
                        (vehicles.lon > -8) & (vehicles.lon <  2)].copy()
    vehicles = vehicles.merge(
        routes_df[["route_id", "route_short_name", "route_type"]],
        on="route_id", how="left")
    vehicles = vehicles.merge(
        trips_df[["trip_id", "shape_id", "direction_id"]].drop_duplicates("trip_id"),
        on="trip_id", how="left")
    vehicles["direction_id"] = vehicles["direction_id"].apply(_norm_dir)
    return vehicles, feed_ts


def resolve_inputs(data_dir, dates, region, time_from=None, time_to=None):
    """Return (bin_path, gtfs_dir, date, utc_offset) tuples for each snapshot."""
    out = []
    for date in dates:
        date_dir   = Path(data_dir) / date
        gtfs_dir   = date_dir / region
        gtfsrt_dir = date_dir / "gtfsrt"
        if not gtfsrt_dir.exists():
            print(f"  [WARN] no gtfsrt for {date}: {gtfsrt_dir}")
            continue
        for bin_path in sorted(gtfsrt_dir.glob("gtfsrt-*.bin")):
            try:
                ts_part = bin_path.stem.split("-", 1)[1]
                file_dt = _dt.datetime.strptime(ts_part, "%Y%m%dT%H%M%S")
                file_ts = int(file_dt.replace(tzinfo=_dt.timezone.utc).timestamp())
            except Exception:
                continue
            utc_off = auto_utc_offset(file_ts)
            local   = _dt.datetime.utcfromtimestamp(file_ts + utc_off)
            sec     = local.hour * 3600 + local.minute * 60 + local.second
            if time_from is not None and sec <  time_from: continue
            if time_to   is not None and sec >= time_to:   continue
            out.append((bin_path, gtfs_dir, date, utc_off))
    return out
```

## 5.4 Route geometry — split by `(route_id, direction_id)`

Same A* + shared-path-cache approach as v3, but every loop iterates over
`(route_id, direction_id)` keys. `route_method` and `route_geoms` use the same
key shape.


```python
import math


def _flat_earth_m(lat1, lon1, lat2, lon2):
    dlat = (lat2 - lat1) * 111_000.0
    dlon = (lon2 - lon1) * 111_000.0 * math.cos(math.radians((lat1 + lat2) * 0.5))
    return math.sqrt(dlat * dlat + dlon * dlon)


def _network_key(route_id, route_type_map):
    rt = route_type_map.get(route_id)
    if rt is None or pd.isna(rt):
        return "drive"
    return ROUTE_TYPE_NETWORK.get(int(rt), "drive")


def _astar_path(G, u, v, cutoff_m):
    nx_v = G.nodes[v]
    gx, gy = nx_v["x"], nx_v["y"]
    def _h(a, b):
        ax, ay = G.nodes[a]["x"], G.nodes[a]["y"]
        return _flat_earth_m(ay, ax, gy, gx)
    path = nx.astar_path(G, u, v, heuristic=_h, weight="length")
    total = 0.0
    for i in range(len(path) - 1):
        ed = G.get_edge_data(path[i], path[i+1])
        if ed:
            total += min(d.get("length", 0.0) or 0.0 for d in ed.values())
        if total > cutoff_m:
            raise nx.NetworkXNoPath(f"path > cutoff {cutoff_m:.0f}m")
    return path


def _precompute_stop_nodes(gtfs, networks):
    """Resolve every (network, stop) → nearest OSM node in one batch per network."""
    from collections import defaultdict
    stop_coords = gtfs["stop_coords"]
    rt_map = gtfs["route_type_map"]
    stops_per_net = defaultdict(set)
    for (rid, did), tids in gtfs["route_dir_trips_map"].items():
        net = _network_key(rid, rt_map)
        for tid in tids:
            for sid in gtfs["trip_stops"].get(tid, []):
                if sid in stop_coords:
                    stops_per_net[net].add(sid)
    out = {}
    for net, sset in stops_per_net.items():
        G = networks.get(net)
        if G is None:
            out[net] = {}
            continue
        sl = list(sset)
        xs = [stop_coords[s][0] for s in sl]
        ys = [stop_coords[s][1] for s in sl]
        print(f"  Batch nearest_nodes ({net}): {len(sl):,} stops …", flush=True)
        try:
            nodes = ox.nearest_nodes(G, X=xs, Y=ys)
        except Exception as e:
            print(f"    fallback per-stop ({e})")
            nodes = [ox.nearest_nodes(G, X=x, Y=y) for x, y in zip(xs, ys)]
        out[net] = dict(zip(sl, nodes))
    return out


def _build_line_osm(stop_ids, G, stop_coords, stop_node_map,
                    path_cache, pair_stats):
    nodes, coords = [], []
    for sid in stop_ids:
        c = stop_coords.get(sid); nn = stop_node_map.get(sid)
        if c is None or nn is None:
            continue
        nodes.append(nn); coords.append(c)
    if len(nodes) < 2:
        raw = [stop_coords[s] for s in stop_ids if s in stop_coords]
        return (LineString(raw), "straight_line") if len(raw) >= 2 else (None, "failed")

    all_coords, routed_ok, gap_filled = [], 0, 0
    for i in range(len(nodes) - 1):
        u, v = nodes[i], nodes[i+1]
        if u == v:
            if not all_coords or all_coords[-1] != coords[i]:
                all_coords.append(coords[i])
            continue
        lon1, lat1 = coords[i]; lon2, lat2 = coords[i+1]
        straight = _flat_earth_m(lat1, lon1, lat2, lon2)
        cutoff   = max(straight * 5.0, 500.0)
        pair_stats["total"] += 1
        seg_key = (id(G), u, v)
        try:
            if seg_key in path_cache:
                seg = path_cache[seg_key]
                if seg is None:
                    raise nx.NetworkXNoPath
                pair_stats["cached"] += 1
            else:
                path = _astar_path(G, u, v, cutoff)
                seg  = [(G.nodes[n]["x"], G.nodes[n]["y"]) for n in path]
                path_cache[seg_key] = seg
                pair_stats["routed"] += 1
            all_coords.extend(seg if not all_coords else seg[1:])
            routed_ok += 1
        except (nx.NetworkXNoPath, nx.NodeNotFound):
            path_cache[seg_key] = None
            a, b = coords[i], coords[i+1]
            if not all_coords:
                all_coords.append(a)
            all_coords.append(b)
            gap_filled += 1
            pair_stats["straight"] += 1

    if len(all_coords) < 2:
        return None, "failed"
    if routed_ok == 0:        method = "straight_line"
    elif gap_filled == 0:     method = "osm_routed"
    else:                     method = "mixed"
    return LineString(all_coords), method


def _build_line_straight(stop_ids, stop_coords):
    coords = [stop_coords[s] for s in stop_ids if s in stop_coords]
    return (LineString(coords), "straight_line") if len(coords) >= 2 else (None, "failed")


def build_all_routes(gtfs, networks, checkpoint_file=None):
    """Build a LineString per (route_id, direction_id)."""
    import time, pickle
    from collections import Counter

    stop_coords    = gtfs["stop_coords"]
    rt_map         = gtfs["route_type_map"]
    rd_canonical   = gtfs["route_dir_canonical"]

    print("  Phase 1: batch nearest_nodes …")
    stop_to_node = _precompute_stop_nodes(gtfs, networks)

    keys = list(rd_canonical.keys())
    n = len(keys)
    route_geoms, route_method = {}, {}

    ckpt = Path(checkpoint_file) if checkpoint_file else None
    if ckpt and ckpt.exists():
        with open(ckpt, "rb") as fh:
            data = pickle.load(fh)
        route_geoms  = data["route_geoms"]
        route_method = data["route_method"]
        keys = [k for k in keys if k not in route_geoms]
        print(f"  Phase 2: resuming — {len(route_geoms)} done, {len(keys)} remain")
    else:
        print(f"  Phase 2: building {n} route×direction LineStrings …")

    path_cache = {}
    pair_stats = {"total": 0, "cached": 0, "routed": 0, "straight": 0}
    REPORT = max(1, n // 20)
    SAVE   = max(1, n // 10)
    t0 = time.monotonic()

    for i, key in enumerate(keys):
        rid, did = key
        stops_ = rd_canonical.get(key, [])
        if len(stops_) < 2:
            route_geoms[key]  = None
            route_method[key] = "no_stops"
        else:
            net = _network_key(rid, rt_map)
            G   = networks.get(net)
            if G is None:
                line, m = _build_line_straight(stops_, stop_coords)
            else:
                line, m = _build_line_osm(stops_, G, stop_coords,
                                          stop_to_node.get(net, {}),
                                          path_cache, pair_stats)
            route_geoms[key]  = line
            route_method[key] = m
        if (i+1) % REPORT == 0 or i+1 == len(keys):
            done = len(route_geoms)
            elapsed = time.monotonic() - t0
            print(f"    {done}/{n}  ({100*done//n}%)  {elapsed:5.0f}s")
        if ckpt and (i+1) % SAVE == 0:
            with open(ckpt, "wb") as fh:
                pickle.dump({"route_geoms": route_geoms,
                             "route_method": route_method}, fh)

    print(f"  Phase 2 done in {time.monotonic()-t0:.1f}s")
    method_counts = Counter(route_method.values())
    for m, c in sorted(method_counts.items(), key=lambda x: -x[1]):
        print(f"    {m:15s}: {c}")
    return route_geoms, route_method
```

## 5.5 Stop projection — keyed by `(route_id, direction_id)`

```python
def project_stops_onto_routes(gtfs, route_geoms, to_metric):
    """Output: {(route_id, direction_id): {stop_id: {frac, dist_m, snap_lon, snap_lat}}}."""
    stop_coords     = gtfs["stop_coords"]
    route_dir_stops = gtfs["route_dir_stops"]
    trans_fwd = Transformer.from_crs(WGS84, METRIC_CRS, always_xy=True)

    out = {}
    for key, stop_ids in route_dir_stops.items():
        line = route_geoms.get(key)
        if line is None or line.length == 0:
            out[key] = {}; continue

        sids, lons, lats = [], [], []
        for sid in stop_ids:
            c = stop_coords.get(sid)
            if c is None:
                continue
            sids.append(sid); lons.append(c[0]); lats.append(c[1])
        if not sids:
            out[key] = {}; continue

        lons = np.asarray(lons); lats = np.asarray(lats)
        xs_m, ys_m = trans_fwd.transform(lons, lats)
        line_m = shp_transform(to_metric, line)
        L_m = line_m.length if line_m.length > 0 else 1.0

        pts_m   = shapely.points(xs_m, ys_m)
        d_along = shapely.line_locate_point(line_m, pts_m)
        snap_m  = shapely.line_interpolate_point(line_m, d_along)
        dist_m  = shapely.distance(pts_m, snap_m)
        fracs   = d_along / L_m

        snap_wgs = shapely.line_interpolate_point(line, fracs, normalized=True)
        snlon = shapely.get_x(snap_wgs); snlat = shapely.get_y(snap_wgs)

        out[key] = {sid: {"frac":     float(fracs[i]),
                          "dist_m":   float(dist_m[i]),
                          "snap_lon": float(snlon[i]),
                          "snap_lat": float(snlat[i])}
                    for i, sid in enumerate(sids)}
    return out
```

## 5.6 Vehicle snapping — group by `(route_id, direction_id)`

```python
def snap_all_vehicles(vehicles, route_geoms, networks, gtfs,
                      to_metric, tol_m=5.0):
    rt_map = gtfs["route_type_map"]
    n = len(vehicles)
    print(f"  Snapping {n:,} pings …")

    route_lines_m = {}
    for key, line in route_geoms.items():
        if line is None or line.length == 0:
            continue
        route_lines_m[key] = shp_transform(to_metric, line)

    snap_lons  = np.full(n, np.nan)
    snap_lats  = np.full(n, np.nan)
    linear_ref = np.full(n, np.nan)
    on_route   = np.zeros(n, dtype=bool)
    dist_m     = np.full(n, np.nan)
    source     = np.array(["none"] * n, dtype=object)

    idx = vehicles.reset_index(drop=True)
    lon_arr = idx["lon"].to_numpy()
    lat_arr = idx["lat"].to_numpy()
    rid_arr = idx["route_id"].to_numpy()
    did_arr = idx["direction_id"].to_numpy()

    # Build (route_id, direction_id) groupby
    rd_keys = list(zip(rid_arr, did_arr))
    rd_arr  = np.array(rd_keys, dtype=object)
    by_rd   = {}
    for i, k in enumerate(rd_keys):
        by_rd.setdefault(k, []).append(i)

    trans_fwd = Transformer.from_crs(WGS84, METRIC_CRS, always_xy=True)
    on_count = off_count = missing = 0
    edge_geom_lookups = {}

    for key, idxs in by_rd.items():
        idxs = np.asarray(idxs, dtype=np.int64)
        line_m = route_lines_m.get(key)
        if line_m is None:
            missing += len(idxs); continue
        L_m = line_m.length if line_m.length > 0 else 1.0
        lons_g = lon_arr[idxs]; lats_g = lat_arr[idxs]
        xs_m, ys_m = trans_fwd.transform(lons_g, lats_g)
        pts_m  = shapely.points(xs_m, ys_m)
        d_al   = shapely.line_locate_point(line_m, pts_m)
        snp_m  = shapely.line_interpolate_point(line_m, d_al)
        d_per  = shapely.distance(pts_m, snp_m)
        dist_m[idxs] = d_per
        on_mask = d_per <= tol_m
        if on_mask.any():
            fracs = d_al[on_mask] / L_m
            wgs   = route_geoms[key]
            snw   = shapely.line_interpolate_point(wgs, fracs, normalized=True)
            on_idxs = idxs[on_mask]
            snap_lons[on_idxs]  = shapely.get_x(snw)
            snap_lats[on_idxs]  = shapely.get_y(snw)
            linear_ref[on_idxs] = fracs
            on_route[on_idxs]   = True
            source[on_idxs]     = "on_route"
            on_count += int(on_mask.sum())
        off_count += int((~on_mask).sum())

    # Off-route → snap to OSM
    to_snap_idx = np.where((~on_route) & pd.notna(rid_arr))[0]
    if len(to_snap_idx):
        net_keys = np.array([_network_key(rid_arr[j], rt_map) for j in to_snap_idx],
                            dtype=object)
        for net in np.unique(net_keys):
            G = networks.get(net)
            sel = to_snap_idx[net_keys == net]
            lons = lon_arr[sel]; lats = lat_arr[sel]
            if G is None:
                snap_lons[sel] = lons; snap_lats[sel] = lats
                source[sel]    = "raw"
                continue
            if net not in edge_geom_lookups:
                lookup = {}
                nx_x = {n_: d.get("x") for n_, d in G.nodes(data=True)}
                nx_y = {n_: d.get("y") for n_, d in G.nodes(data=True)}
                for u, v, kk, data in G.edges(keys=True, data=True):
                    g = data.get("geometry")
                    if g is None:
                        ux, uy = nx_x.get(u), nx_y.get(u)
                        vx, vy = nx_x.get(v), nx_y.get(v)
                        if None in (ux, uy, vx, vy):
                            continue
                        g = LineString([(ux, uy), (vx, vy)])
                    lookup[(u, v, kk)] = g
                edge_geom_lookups[net] = lookup
            lookup = edge_geom_lookups[net]
            try:
                nearest = ox.nearest_edges(G, X=lons, Y=lats)
            except Exception as exc:
                print(f"    [WARN] nearest_edges {net}: {exc}")
                snap_lons[sel] = lons; snap_lats[sel] = lats
                source[sel]    = "raw"; continue
            geoms = [lookup.get(t) for t in nearest]
            has = np.array([g is not None for g in geoms], dtype=bool)
            if has.any():
                pts = shapely.points(lons[has], lats[has])
                lines = np.asarray([g for g in geoms if g is not None], dtype=object)
                d_al  = shapely.line_locate_point(lines, pts)
                snp   = shapely.line_interpolate_point(lines, d_al)
                snap_lons[sel[has]] = shapely.get_x(snp)
                snap_lats[sel[has]] = shapely.get_y(snp)
            if (~has).any():
                snap_lons[sel[~has]] = lons[~has]
                snap_lats[sel[~has]] = lats[~has]
            source[sel] = "osm"

    miss = np.isnan(snap_lons)
    if miss.any():
        snap_lons[miss] = lon_arr[miss]
        snap_lats[miss] = lat_arr[miss]

    idx["snapped_lon"] = snap_lons
    idx["snapped_lat"] = snap_lats
    idx["linear_ref"]  = linear_ref
    idx["on_route"]    = on_route
    idx["dist_m"]      = dist_m
    idx["source"]      = source
    print(f"    on_route: {on_count:,}  |  off_route→osm: {off_count:,}  "
          f"|  no route line: {missing:,}")
    return idx
```

## 5.7 Arrival interpolation — with pre / post extrapolation

The vectorised `_interpolate_trip_vectorised` now also fills in:

* **`extrapolated_pre`** — stop sits before the earliest on-route ping but we
  have ≥ 2 after-pings. Uses velocity = `(f₂−f₁)/(t₂−t₁)` of the first two
  pings (frac-sorted) and extrapolates backward in time.
* **`extrapolated_post`** — stop sits after the latest on-route ping but we
  have ≥ 2 before-pings. Same idea, forward extrapolation from the last two.

Extrapolation is rejected when the inferred velocity is `≤ 0` (bus going
backwards in frac space) or when the gap is unreasonably large.


```python
MAX_EXTRAPOLATE_SEC = 600.0   # don't extrapolate more than 10 min beyond the last/first ping


def _interpolate_trip_vectorised(frac, ts, vids, stop_fracs, diversion_gap_sec):
    """Vectorised interpolation + pre/post extrapolation for one trip."""
    N = len(stop_fracs)
    P = len(frac)

    arrival_sec  = np.full(N, np.nan)
    veh_id       = np.empty(N, dtype=object); veh_id[:] = ""
    sources      = np.empty(N, dtype=object); sources[:] = "no_data"
    n_used       = np.zeros(N, dtype=np.int32)

    if P == 0:
        return {"arrival_sec": arrival_sec, "vehicle_id": veh_id,
                "source": sources, "n_pings_used": n_used}

    order = np.argsort(frac, kind="stable")
    f_s = frac[order]; t_s = ts[order]; v_s = vids[order]
    r   = np.searchsorted(f_s, stop_fracs, side="left")
    has_after  = r < P
    has_before = r > 0

    # ── only-before stops: extrapolate forward if ≥2 before-pings ───────────
    mask_only_before = has_before & ~has_after
    if mask_only_before.any():
        idxs = np.where(mask_only_before)[0]
        for k in idxs:
            ib = r[k] - 1                       # last before ping
            if ib >= 1:                         # have ib-1 too — extrapolate
                f_a = f_s[ib]; f_b = f_s[ib-1]
                t_a = t_s[ib]; t_b = t_s[ib-1]
                df = f_a - f_b; dt = t_a - t_b
                if df > 0 and dt > 0:
                    v = df / dt
                    est = t_a + (stop_fracs[k] - f_a) / v
                    if est - t_a <= MAX_EXTRAPOLATE_SEC:
                        arrival_sec[k] = est
                        sources[k]     = "extrapolated_post"
                        veh_id[k]      = str(v_s[ib])
                        n_used[k]      = 2
                        continue
            sources[k] = "no_data"
            veh_id[k]  = str(v_s[ib])
            n_used[k]  = 1

    # ── only-after stops: extrapolate backward if ≥2 after-pings ────────────
    mask_only_after = has_after & ~has_before
    if mask_only_after.any():
        idxs = np.where(mask_only_after)[0]
        for k in idxs:
            ia = r[k]                           # first after ping
            if ia + 1 < P:                      # have ia+1 too — extrapolate
                f_a = f_s[ia]; f_b = f_s[ia+1]
                t_a = t_s[ia]; t_b = t_s[ia+1]
                df = f_b - f_a; dt = t_b - t_a
                if df > 0 and dt > 0:
                    v = df / dt
                    est = t_a - (f_a - stop_fracs[k]) / v
                    if t_a - est <= MAX_EXTRAPOLATE_SEC:
                        arrival_sec[k] = est
                        sources[k]     = "extrapolated_pre"
                        veh_id[k]      = str(v_s[ia])
                        n_used[k]      = 2
                        continue
            sources[k] = "pre_service"
            veh_id[k]  = str(v_s[ia])
            n_used[k]  = 1

    # ── both sides: standard linear interpolation / dwell / off-route ──────
    mask_both = has_before & has_after
    if mask_both.any():
        ib = r[mask_both] - 1; ia = r[mask_both]
        f_b = f_s[ib]; f_a = f_s[ia]
        t_b = t_s[ib]; t_a = t_s[ia]
        closer_before = np.abs(stop_fracs[mask_both] - f_b) <= np.abs(f_a - stop_fracs[mask_both])
        closer_idx = np.where(closer_before, ib, ia)
        closer_vid = v_s[closer_idx].astype(str)

        off_route = (t_a - t_b) > diversion_gap_sec
        dwell     = (~off_route) & (np.abs(f_a - f_b) < 1e-9)
        normal    = ~off_route & ~dwell
        denom = np.where(normal, (f_a - f_b), 1.0)
        ratio = np.where(normal, (stop_fracs[mask_both] - f_b) / denom, 0.0)
        arr_interp = t_b + ratio * (t_a - t_b)
        arr_dwell  = 0.5 * (t_b + t_a)
        arr = np.full(mask_both.sum(), np.nan)
        arr[normal] = arr_interp[normal]
        arr[dwell]  = arr_dwell[dwell]

        src = np.where(off_route, "off_route",
              np.where(dwell, "dwell", "interpolated"))

        arrival_sec[mask_both] = arr
        sources[mask_both]     = src
        veh_id[mask_both]      = closer_vid
        n_used[mask_both]      = 2

    return {"arrival_sec": arrival_sec, "vehicle_id": veh_id,
            "source": sources, "n_pings_used": n_used}


def compute_all_arrivals(snapped, gtfs, stop_proj, utc_offset, diversion_gap=300.0):
    stop_times      = gtfs["stop_times"]
    stop_name_map   = gtfs["stop_name_map"]
    route_short_map = gtfs["route_short_map"]
    trips_df        = gtfs["trips_df"]

    active = set(snapped["trip_id"].dropna().unique())
    if not active:
        print("  No active trips — empty arrivals.")
        return pd.DataFrame()

    st = stop_times[stop_times["trip_id"].isin(active)].copy()
    meta = trips_df[["trip_id", "route_id", "direction_id"]].drop_duplicates("trip_id")
    st = st.merge(meta, on="trip_id", how="left")
    st["direction_id"] = st["direction_id"].apply(_norm_dir)

    # stop_frac lookup keyed by (route_id, direction_id, stop_id)
    rows = [(rid, did, sid, info["frac"])
            for (rid, did), by_sid in stop_proj.items()
            for sid, info in by_sid.items()]
    if rows:
        frac_df = pd.DataFrame(rows, columns=["route_id", "direction_id",
                                              "stop_id", "stop_frac"])
        st = st.merge(frac_df, on=["route_id", "direction_id", "stop_id"], how="left")
    else:
        st["stop_frac"] = np.nan

    st["stop_name"]        = st["stop_id"].map(stop_name_map)
    st["route_short_name"] = st["route_id"].map(route_short_map)

    on_pings = (snapped[snapped["on_route"]]
                .dropna(subset=["linear_ref", "timestamp"]))
    trip_vid_fallback = (on_pings.groupby("trip_id")["vehicle_id"]
                                 .agg(lambda s: s.value_counts().idxmax()
                                                if len(s) else "")
                                 .to_dict())
    pings_by_trip = {tid: (g["linear_ref"].to_numpy(),
                            g["timestamp"].to_numpy(),
                            g["vehicle_id"].to_numpy())
                     for tid, g in on_pings.groupby("trip_id", sort=False)}

    N = len(st)
    actual_sec = np.full(N, np.nan)
    veh_ids    = np.empty(N, dtype=object); veh_ids[:] = ""
    sources    = np.empty(N, dtype=object); sources[:] = "no_data"
    n_pings    = np.zeros(N, dtype=np.int32)

    stop_frac_arr = st["stop_frac"].to_numpy(dtype=float)
    no_geom = np.isnan(stop_frac_arr)
    sources[no_geom] = "no_route_geom"

    indices_by_trip = st.groupby("trip_id", sort=False).indices
    n_trips = len(indices_by_trip)
    for tn, (tid, idx) in enumerate(indices_by_trip.items(), start=1):
        idx = np.asarray(idx, dtype=np.int64)
        good = ~no_geom[idx]
        if not good.any():
            continue
        idx_g = idx[good]
        sf = stop_frac_arr[idx_g]
        p = pings_by_trip.get(tid)
        if p is None:
            veh_ids[idx_g] = trip_vid_fallback.get(tid, "")
            continue
        f, t, v = p
        res = _interpolate_trip_vectorised(f, t, v, sf, diversion_gap)
        actual_sec[idx_g] = res["arrival_sec"]
        sources[idx_g]    = res["source"]
        n_pings[idx_g]    = res["n_pings_used"]
        vo = res["vehicle_id"]
        empty = (vo == "")
        if empty.any():
            vo = vo.copy(); vo[empty] = trip_vid_fallback.get(tid, "")
        veh_ids[idx_g] = vo
        if tn % 500 == 0 or tn == n_trips:
            print(f"    {tn:,}/{n_trips:,} trips interpolated")

    # Scheduled-epoch anchoring (per service date, not calendar date)
    anchor = (snapped.dropna(subset=["timestamp"])
                       .groupby("trip_id")["timestamp"].median().astype(np.int64))
    first_arr = (st.dropna(subset=["arr_sec"])
                   .groupby("trip_id")["arr_sec"].min())
    anc_ts  = st["trip_id"].map(anchor).to_numpy(dtype=float)
    fa_sec  = st["trip_id"].map(first_arr).to_numpy(dtype=float)
    arr_sec = st["arr_sec"].to_numpy(dtype=float)
    SECS_PER_DAY = 86400
    local_ts = anc_ts + utc_offset - fa_sec
    local_mid = np.floor(local_ts / SECS_PER_DAY) * SECS_PER_DAY
    sched_epoch = (local_mid - utc_offset) + arr_sec
    sched_epoch[np.isnan(anc_ts) | np.isnan(arr_sec) | np.isnan(fa_sec)] = np.nan

    st["scheduled_epoch"] = sched_epoch
    st["actual_sec"]      = actual_sec
    st["vehicle_id"]      = veh_ids
    st["arrival_source"]  = sources
    st["n_pings_used"]    = n_pings
    delay = st["actual_sec"] - st["scheduled_epoch"]
    st["delay_sec"] = delay
    st["delay_min"] = (delay / 60).round(2)

    def _fmt_epoch(e, off):
        out = np.empty(len(e), dtype=object); out[:] = ""
        m = ~np.isnan(e)
        if m.any():
            l = (e[m] + off).astype(np.int64)
            hh = (l // 3600) % 24; mm = (l // 60) % 60; ss = l % 60
            f = np.char.add(np.char.zfill(hh.astype(str), 2), ":")
            f = np.char.add(f, np.char.zfill(mm.astype(str), 2))
            f = np.char.add(f, ":")
            f = np.char.add(f, np.char.zfill(ss.astype(str), 2))
            out[m] = f
        return out

    st["scheduled_arrival"] = _fmt_epoch(st["scheduled_epoch"].to_numpy(dtype=float),
                                         utc_offset)
    st["actual_arrival"]    = _fmt_epoch(st["actual_sec"].to_numpy(dtype=float),
                                         utc_offset)

    cols = ["route_id", "route_short_name", "trip_id", "vehicle_id",
            "direction_id", "stop_id", "stop_name", "stop_sequence",
            "stop_frac", "scheduled_arrival", "actual_arrival",
            "delay_sec", "delay_min", "arrival_source", "n_pings_used"]
    for c in cols:
        if c not in st.columns:
            st[c] = np.nan
    arrivals = st[cols].sort_values(
        ["route_id", "direction_id", "trip_id", "stop_sequence"]
    ).reset_index(drop=True)

    from collections import Counter
    print("  Arrival sources:")
    for s, c in Counter(arrivals["arrival_source"]).most_common():
        print(f"    {s:18s}: {c:,}")
    return arrivals
```

## 5.8 Output builders (`route_id` × `direction_id`)

```python
def build_gdf_routes(route_geoms, route_method, gtfs, arrivals):
    rt_map = gtfs["route_type_map"]
    rs_map = gtfs["route_short_map"]
    if len(arrivals):
        nstops = (arrivals.groupby(["route_id", "direction_id"])["stop_id"]
                          .nunique().to_dict())
        ntrips = (arrivals.groupby(["route_id", "direction_id"])["trip_id"]
                          .nunique().to_dict())
    else:
        nstops, ntrips = {}, {}
    rows = []
    for (rid, did), line in route_geoms.items():
        if line is None:
            continue
        rt = rt_map.get(rid)
        rt_name = ROUTE_TYPE_NAMES.get(int(rt), f"type{rt}") if pd.notna(rt) else "unknown"
        rows.append({
            "route_id":         rid,
            "direction_id":     did,
            "route_short_name": rs_map.get(rid, ""),
            "route_type":       int(rt) if pd.notna(rt) else -1,
            "route_type_name":  rt_name,
            "geometry_method":  route_method.get((rid, did), "unknown"),
            "n_stops_covered":  int(nstops.get((rid, did), 0)),
            "n_trips_observed": int(ntrips.get((rid, did), 0)),
            "geometry":         line,
        })
    return gpd.GeoDataFrame(rows, geometry="geometry", crs=WGS84)


def build_gdf_stops(stop_proj, gtfs, arrivals):
    stop_name_map = gtfs["stop_name_map"]
    rs_map        = gtfs["route_short_map"]
    if len(arrivals):
        agg = (arrivals.groupby(["route_id", "direction_id", "stop_id"])
                       .agg(n_trips=("trip_id", "nunique"),
                            n_actual=("actual_arrival",
                                      lambda s: int((s != "").sum())),
                            avg_delay_min=("delay_min", "mean"),
                            sources=("arrival_source",
                                     lambda s: " | ".join(sorted(set(s)))))
                       .reset_index())
        agg["avg_delay_min"] = agg["avg_delay_min"].round(2)
    else:
        agg = pd.DataFrame(columns=["route_id", "direction_id", "stop_id",
                                    "n_trips", "n_actual",
                                    "avg_delay_min", "sources"])

    rids, dids, sids, fracs, dists, lons, lats = [], [], [], [], [], [], []
    for (rid, did), stops in stop_proj.items():
        for sid, info in stops.items():
            rids.append(rid); dids.append(did); sids.append(sid)
            fracs.append(info["frac"]); dists.append(info["dist_m"])
            lons.append(info["snap_lon"]); lats.append(info["snap_lat"])

    if not rids:
        return gpd.GeoDataFrame(columns=["route_id", "direction_id", "stop_id",
                                          "stop_name", "stop_frac",
                                          "dist_to_route_m", "geometry"],
                                 geometry="geometry", crs=WGS84)

    df = pd.DataFrame({
        "route_id":         rids,
        "direction_id":     dids,
        "route_short_name": pd.Series(rids).map(rs_map).fillna("").values,
        "stop_id":          sids,
        "stop_name":        pd.Series(sids).map(stop_name_map).fillna("").values,
        "stop_frac":        np.asarray(fracs, dtype=float),
        "dist_to_route_m":  np.asarray(dists, dtype=float),
    })
    geom = gpd.points_from_xy(lons, lats)
    gdf = gpd.GeoDataFrame(df, geometry=geom, crs=WGS84)
    if len(agg):
        gdf = gdf.merge(agg, on=["route_id", "direction_id", "stop_id"], how="left")
    return gdf


def build_gdf_vehicles(snapped):
    if len(snapped) == 0:
        return gpd.GeoDataFrame(geometry=[], crs=WGS84)
    cols = ["trip_id", "route_id", "route_short_name", "vehicle_id",
            "timestamp", "on_route", "source", "dist_m", "linear_ref",
            "direction_id", "bearing"]
    cols = [c for c in cols if c in snapped.columns]
    geom = gpd.points_from_xy(snapped["snapped_lon"], snapped["snapped_lat"])
    return gpd.GeoDataFrame(snapped[cols].copy(), geometry=geom, crs=WGS84)


def write_output(out_path, gdf_routes, gdf_stops, gdf_vehicles, arrivals,
                 write_csv=True):
    out_path = Path(out_path)
    out_path.parent.mkdir(parents=True, exist_ok=True)
    if "on_route" in gdf_vehicles.columns:
        gdf_vehicles = gdf_vehicles.copy()
        gdf_vehicles["on_route"] = gdf_vehicles["on_route"].astype("int8")
    if out_path.exists():
        out_path.unlink()
    gdf_routes.to_file(out_path,   layer="routes",   driver="GPKG")
    gdf_stops.to_file(out_path,    layer="stops",    driver="GPKG", mode="a")
    gdf_vehicles.to_file(out_path, layer="vehicles", driver="GPKG", mode="a")
    print(f"  GPKG → {out_path}  "
          f"(routes={len(gdf_routes)}, stops={len(gdf_stops)}, "
          f"vehicles={len(gdf_vehicles)})")
    if write_csv:
        csv = out_path.with_suffix(".arrivals.csv")
        arrivals.to_csv(csv, index=False)
        print(f"  CSV  → {csv}  ({len(arrivals):,} rows)")
```

## 5.9 Run the pipeline (unified-static driver)

The driver runs in **two stages**:

### Stage A — Build shared static state (once, across ALL dates)
1. Stream-load each date's static GTFS, accumulating a **union**:
   `stop_coords`, `trip_stops`, `route_dir_trips_map`, etc.
2. Hash each date's GTFS and report consistency (✓ identical / ⚠ differs).
3. **Snap stops to OSM nodes ONCE** for the unioned stop set — this is the
   most expensive step and re-running it per date was the previous bottleneck.
4. Run **A\* route geometry construction ONCE** for every `(route_id,
   direction_id)` key in the union. Cached to
   `output/cache/route_geoms_v3_dir_unified.pickle`.
5. Project every union stop onto its route line.

### Stage B — Per-date RT processing
For each date in `DATES`:
1. Load the date's `stop_times.txt` / `trips.txt` (date-specific schedules).
2. Read GTFS-RT `.bin` files with running-progress reporting.
3. Snap pings to the **shared** route geometries.
4. Interpolate / extrapolate arrivals using the **shared** stop projection.
5. Write `output/routes_<date>.gpkg` + `.arrivals.csv`, free per-date frames.

The shared cache means the second run for the same date set is essentially
just the per-date RT loop.


```python
import time as _t
import hashlib as _hashlib
import pickle as _pickle


def _print_step(msg, char="="):
    print(f"\n{char*60}\n{msg}\n{char*60}", flush=True)


def _gtfs_quick_hash(gtfs) -> str:
    """Capture (stops × trip→stops × route_dir) shape of a GTFS feed."""
    h = _hashlib.md5()
    for sid in sorted(gtfs["stop_coords"]):
        h.update(sid.encode())
    for tid in sorted(gtfs["trip_stops"]):
        h.update(tid.encode())
        h.update("|".join(gtfs["trip_stops"][tid]).encode())
    for key in sorted(gtfs["route_dir_trips_map"]):
        h.update(str(key).encode())
    return h.hexdigest()[:10]


# ─────────────────────────────────────────────────────────────────────────────
# STAGE A — shared static state
# ─────────────────────────────────────────────────────────────────────────────

def build_shared_static_state(dates, data_dir, region, networks,
                              route_id_filter, geom_cache_path):
    """Build the route-geom + stop-projection state ONCE across all dates."""
    _print_step("STAGE A — Build shared static state across all dates")

    # ── A.1  Stream-load each date, accumulate union, hash for consistency
    print(f"\n[A.1/3] Loading static GTFS for {len(dates)} date(s) ...")
    U_stops, U_name        = {}, {}
    U_rt, U_short          = {}, {}
    U_rd_trips             = {}     # (rid, did) → set[tid]
    U_trip_stops           = {}     # tid → list[sid]  (longest seen)
    U_trip_dir             = {}     # tid → did
    per_date_hash          = {}

    for di, d in enumerate(dates, start=1):
        gtfs_dir = data_dir / d / region
        if not gtfs_dir.is_dir():
            print(f"   [{di}/{len(dates)}] {d} — folder missing, skipping")
            continue
        t0 = _t.monotonic()
        g = load_static_gtfs(gtfs_dir, route_id_filter=route_id_filter)

        for sid, c in g["stop_coords"].items():    U_stops.setdefault(sid, c)
        for sid, n in g["stop_name_map"].items():  U_name.setdefault(sid, n)
        for rid, rt in g["route_type_map"].items():  U_rt[rid] = rt
        for rid, rs in g["route_short_map"].items(): U_short[rid] = rs
        for key, tids in g["route_dir_trips_map"].items():
            U_rd_trips.setdefault(key, set()).update(tids)
        for tid, sids in g["trip_stops"].items():
            ex = U_trip_stops.get(tid)
            if ex is None or len(sids) > len(ex):
                U_trip_stops[tid] = sids
        for tid, did in g["trip_dir"].items():
            U_trip_dir.setdefault(tid, did)

        per_date_hash[d] = _gtfs_quick_hash(g)
        print(f"   [{di}/{len(dates)}] {d}  "
              f"stops={len(g['stops_df']):>6,}  "
              f"trips={len(g['trips_df']):>7,}  "
              f"hash={per_date_hash[d]}  "
              f"({_t.monotonic()-t0:5.1f}s)")
        del g  # free immediately

    # Consistency report
    if len(set(per_date_hash.values())) == 1:
        print(f"\n   ✓ Static GTFS identical across all {len(per_date_hash)} dates "
              f"(hash {next(iter(per_date_hash.values()))})")
        consistent = True
    else:
        print(f"\n   ⚠ Static GTFS DIFFERS across dates — using union of all stops/trips:")
        for d, h in per_date_hash.items():
            print(f"        {d}: hash {h}")
        consistent = False

    print(f"\n   Union → {len(U_stops):>6,} stops · "
          f"{len(U_trip_stops):>7,} trips · "
          f"{len(U_rd_trips):>5,} (route, dir) keys")

    # ── A.2  Derive route_dir_stops + route_dir_canonical from union
    U_rd_stops, U_rd_canonical = {}, {}
    for key, tids in U_rd_trips.items():
        seen, seen_set = [], set()
        canon = []
        for tid in tids:
            ts = U_trip_stops.get(tid, [])
            if len(ts) > len(canon):
                canon = ts
            for sid in ts:
                if sid not in seen_set:
                    seen.append(sid); seen_set.add(sid)
        U_rd_stops[key]     = seen
        U_rd_canonical[key] = canon

    shared_gtfs = {
        "stop_coords":         U_stops,
        "stop_name_map":       U_name,
        "route_type_map":      U_rt,
        "route_short_map":     U_short,
        "route_dir_trips_map": {k: list(v) for k, v in U_rd_trips.items()},
        "route_dir_canonical": U_rd_canonical,
        "route_dir_stops":     U_rd_stops,
        "trip_stops":          U_trip_stops,
        "trip_dir":            U_trip_dir,
    }

    # ── A.3  Route LineStrings — cached pickle keyed to the union
    print(f"\n[A.2/3] Building route LineStrings "
          f"({len(U_rd_canonical):,} (route, dir) keys) ...")
    if geom_cache_path.exists():
        size_mb = geom_cache_path.stat().st_size / 1_048_576
        print(f"   Cache hit ({size_mb:.0f} MB) — loading ...", flush=True)
        t0 = _t.monotonic()
        with open(geom_cache_path, "rb") as fh:
            data = _pickle.load(fh)
        route_geoms  = data["route_geoms"]
        route_method = data["route_method"]
        print(f"   {len(route_geoms):,} geoms loaded ({_t.monotonic()-t0:.1f}s)")
    else:
        ckpt = geom_cache_path.with_suffix(".checkpoint.pickle")
        route_geoms, route_method = build_all_routes(
            shared_gtfs, networks, checkpoint_file=ckpt)
        geom_cache_path.parent.mkdir(parents=True, exist_ok=True)
        with open(geom_cache_path, "wb") as fh:
            _pickle.dump({"route_geoms": route_geoms,
                          "route_method": route_method}, fh)
        ckpt.unlink(missing_ok=True)
        print(f"   Cached → {geom_cache_path.name}")

    # ── A.4  Stop projection (vectorised, fast)
    print(f"\n[A.3/3] Projecting stops onto routes ...")
    t0 = _t.monotonic()
    to_metric, _ = make_projectors()
    stop_proj = project_stops_onto_routes(shared_gtfs, route_geoms, to_metric)
    n_proj = sum(len(v) for v in stop_proj.values())
    print(f"   {n_proj:,} (route, dir, stop) projections ({_t.monotonic()-t0:.1f}s)")

    return {
        **shared_gtfs,
        "route_geoms":   route_geoms,
        "route_method":  route_method,
        "stop_proj":     stop_proj,
        "to_metric":     to_metric,
        "per_date_hash": per_date_hash,
        "consistent":    consistent,
    }


# ─────────────────────────────────────────────────────────────────────────────
# STAGE B — per-date RT processing
# ─────────────────────────────────────────────────────────────────────────────

def process_date(date, shared, data_dir, region, networks, output_dir,
                 route_id_filter):
    """Load RT pings, snap, compute arrivals, write output for one date."""
    _print_step(f"DATE {date}", char="-")
    t_total = _t.monotonic()

    # B.1 Snapshots
    snapshots = resolve_inputs(data_dir, [date], region)
    if not snapshots:
        print(f"   no snapshots — skipping"); return None
    print(f"   [1/5] {len(snapshots)} snapshots discovered")

    # B.2 Date-specific schedule
    t0 = _t.monotonic()
    date_gtfs = load_static_gtfs(snapshots[0][1], route_id_filter=route_id_filter)
    # Inject shared metadata so downstream functions see consistent maps
    date_gtfs["stop_name_map"]   = shared["stop_name_map"]
    date_gtfs["route_short_map"] = shared["route_short_map"]
    date_gtfs["route_type_map"]  = shared["route_type_map"]
    print(f"   [2/5] schedule loaded ({_t.monotonic()-t0:.1f}s)")

    # B.3 RT pings — sequential read with running progress
    t0 = _t.monotonic()
    pings = []
    n_total = len(snapshots)
    for i, (bp, _, _, _) in enumerate(snapshots, start=1):
        v, _ = load_gtfsrt(bp, date_gtfs["gtfs_trip_ids"],
                           date_gtfs["routes_df"], date_gtfs["trips_df"])
        if v is not None:
            pings.append(v)
        if i % 100 == 0 or i == n_total:
            n_pings = sum(len(p) for p in pings)
            print(f"   [3/5] RT {i:>5d}/{n_total}  "
                  f"{n_pings:>9,} pings  "
                  f"({_t.monotonic()-t0:5.1f}s)", end="\r", flush=True)
    print()
    if not pings:
        print(f"   no RT data — skipping"); return None
    vehicles = pd.concat(pings, ignore_index=True)
    del pings
    print(f"         {len(vehicles):,} pings, "
          f"{vehicles['trip_id'].nunique():,} trips, "
          f"{vehicles['route_id'].nunique():,} routes")

    # B.4 Snap
    t0 = _t.monotonic()
    print(f"   [4/5] snapping vehicles ...")
    snapped = snap_all_vehicles(vehicles, shared["route_geoms"], networks,
                                date_gtfs, shared["to_metric"], tol_m=5.0)
    del vehicles
    print(f"         done ({_t.monotonic()-t0:.1f}s)")

    # B.5 Arrivals
    t0 = _t.monotonic()
    print(f"   [5/5] computing arrivals ...")
    arrivals = compute_all_arrivals(snapped, date_gtfs, shared["stop_proj"],
                                    utc_offset=snapshots[0][3],
                                    diversion_gap=300.0)
    print(f"         done ({_t.monotonic()-t0:.1f}s)")

    # B.6 Output
    t0 = _t.monotonic()
    out_path = output_dir / f"routes_{date}.gpkg"
    gdf_routes   = build_gdf_routes(shared["route_geoms"], shared["route_method"],
                                     shared, arrivals)
    gdf_stops    = build_gdf_stops(shared["stop_proj"], shared, arrivals)
    gdf_vehicles = build_gdf_vehicles(snapped)
    write_output(out_path, gdf_routes, gdf_stops, gdf_vehicles, arrivals)
    del snapped, gdf_vehicles
    print(f"   [{date}] complete in {_t.monotonic()-t_total:.1f}s")
    return arrivals


# ─────────────────────────────────────────────────────────────────────────────
# DRIVER — Stage A then Stage B for each date in DATES
# ─────────────────────────────────────────────────────────────────────────────

ROUTE_GEOM_CACHE_DIR = OUTPUT_DIR / "cache"
ROUTE_GEOM_CACHE_DIR.mkdir(parents=True, exist_ok=True)
SHARED_GEOM_CACHE = ROUTE_GEOM_CACHE_DIR / "route_geoms_v3_dir_unified.pickle"

shared = build_shared_static_state(
    DATES, DATA_DIR, REGION, networks,
    selected_route_ids, SHARED_GEOM_CACHE,
)

_print_step(f"STAGE B — Process {len(DATES)} date(s)")
all_arrivals = {}
for date in DATES:
    a = process_date(date, shared, DATA_DIR, REGION, networks,
                     OUTPUT_DIR, selected_route_ids)
    if a is not None:
        all_arrivals[date] = a

_print_step("ALL DONE")
print(f"  processed {len(all_arrivals)}/{len(DATES)} dates")
for d, a in all_arrivals.items():
    print(f"    {d}: {len(a):,} arrival rows")
```
